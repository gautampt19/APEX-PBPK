```
# Interactive Model Parameter Loader and Validator
def get_model_parameters(model_key):
    csv_file = f"{model_key}_parameters.csv"
    if not os.path.exists(csv_file):
        raise FileNotFoundError(f"Cannot find {csv_file}")
    df = pd.read_csv(csv_file)
    print(f"Loaded {model_key} with {len(df)} total parameters.")
    return df

# Example: Inspect all parameters for Pesticides_Zhao_2023
df_zhao = get_model_parameters("Pesticides_Zhao_2023")
df_zhao.head(15)
```

```
# Visualizing Parameter Verification Distribution by Chemical Group
plt.figure(figsize=(12, 6))
group_summary = df_summary.groupby('Group')[['Exact_Matches', 'Formulas', 'Ref_Standards']].sum()
group_summary.plot(kind='bar', stacked=True, figsize=(12, 6), color=['#2ca02c', '#1f77b4', '#ff7f0e'])
plt.title("Parameter Verification Breakdown Across Chemical Classes", fontsize=14, fontweight='bold')
plt.xlabel("Chemical Group", fontsize=12)
plt.ylabel("Number of Model Parameters", fontsize=12)
plt.legend(["Exact Literature Matches", "Mathematical Rules / Formulas", "Reference Physiological Standards"])
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
